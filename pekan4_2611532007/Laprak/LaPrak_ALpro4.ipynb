## 2.1 Konsep Dasar: Logika Percabangan pada Pemrograman

Pengambilan keputusan dalam dunia komputasi sangat bergantung pada struktur kontrol percabangan (selection/decision structure). Sebagai salah satu fondasi algoritma—bersanding dengan urutan, perulangan, dan subrutin—percabangan memungkinkan kode komputer untuk tidak sekadar berjalan lurus, melainkan mampu beradaptasi dan memilih rute tindakan berdasarkan kondisi data saat itu. Dengan kata lain, fitur ini menjadi kunci yang menyulap sekumpulan instruksi kaku menjadi sebuah sistem yang pintar dan responsif.

Mengacu pada literatur pendidikan komputasi (computing education), seberapa baik seorang pemula menguasai konsep percabangan sering kali menentukan keberhasilannya belajar pemrograman. Menurut Ebrahimi (1994), kesulitan yang umum dialami terbagi menjadi dua ranah: kendala pada penulisan sintaksis (language constructs) dan kebingungan dalam merangkai alur logika (plan composition). Sejalan dengan hal tersebut, Green (1977) menegaskan bahwa gaya penulisan blok kondisional—termasuk penempatan negasi dan alternatif cabang—memiliki dampak langsung terhadap kemudahan membaca kode dan tingkat kelelahan mental pemrogram saat harus memperbaiki atau memelihara kode tersebut.

### 2.1.1 Pernyataan IF Tunggal (Filter Sederhana)

Struktur if dasar berperan layaknya sebuah pos pemeriksaan untuk sekelompok instruksi. Instruksi-instruksi di dalamnya baru akan dijalankan apabila sebuah kriteria tunggal berhasil dipenuhi. Logikanya sangat lugas: jika pengecekan menghasilkan nilai Benar (True), komputer akan mengeksekusi blok tersebut; sebaliknya, jika Salah (False), komputer akan melewatinya begitu saja dan melanjutkan ke perintah berikutnya. Model ini sangat pas diterapkan untuk pengujian satu arah, misalnya sekadar untuk menampilkan peringatan error tanpa harus mengatur tindakan spesifik jika semuanya berjalan normal.

### 2.1.2 IF Bersarang (Logika Berlapis)

Kondisi nested if atau IF bersarang terjadi manakala kita memasukkan sebuah struktur percabangan ke dalam perut percabangan lain, sehingga menghasilkan syarat yang berjenjang. Pada skenario ini, nasib dari blok kondisi terdalam (inner if) sepenuhnya disetir oleh kondisi terluar (outer if). Jika syarat di pintu paling luar sudah gagal dipenuhi, maka seluruh kode di bagian dalam otomatis tidak akan pernah dibaca apalagi dieksekusi oleh mesin. Praktik pembuatan syarat berlapis ini lazim ditemui pada kasus-kasus yang menuntut pengecekan data, klasifikasi bertingkat, atau validasi keamanan secara bertahap.

### 2.1.3 Multi IF (Evaluasi Syarat Independen)

Walaupun terlihat berderet, Multi IF sejatinya adalah sekumpulan blok if terpisah yang dievaluasi sendiri-sendiri oleh komputer. Mesin akan mengecek setiap syarat yang ada dari atas ke bawah tanpa pengecualian, tidak peduli apakah syarat di atasnya sudah terpenuhi atau belum. Sifat penumpukan atau evaluasi akumulatif ini membuat Multi IF menjadi solusi paling ideal untuk kasus di mana beberapa ketentuan bisa terjadi di waktu yang sama (tidak saling eksklusif), sehingga setiap syarat yang benar dapat ditangani secara berurutan.

### 2.1.4 IF-ELSE (Bifurkasi Tindakan)

Pola IF-ELSE menciptakan persimpangan dua arah yang mengikat, di mana program pasti akan menempuh salah satu jalur. Karena kedua opsi ini sifatnya berlawanan dan saling menggugurkan, program terhindar dari situasi "menggantung" tanpa adanya eksekusi sama sekali. Seperti yang disoroti oleh Green (1977), kehadiran blok else secara tertulis sangat meringankan kerja otak saat membaca kode. Pemrogram menjadi lebih yakin dan tidak perlu menerka-nerka apa yang akan dilakukan sistem apabila syarat utama gagal, karena langkah cadangannya sudah ditetapkan secara eksplisit.

### 2.1.5 IF-ELIF-ELSE (Rantai Kondisi Eksklusif)

Ketika dihadapkan pada banyak opsi yang saling menggugurkan, struktur IF-ELIF-ELSE adalah jawabannya. Berlainan dengan Multi IF yang memeriksa semua blok, rangkaian elif dirancang untuk mencari kebenaran pertama saja. Segera setelah ada satu kondisi yang berstatus True, blok tersebut dieksekusi dan semua pengecekan untuk kondisi sisanya akan langsung diabaikan (short-circuit). Pada bagian paling bawah, else disiapkan sebagai pelampung keselamatan (fallback) yang akan menangkap semua kemungkinan lain yang tidak terjaring oleh spesifikasi kondisi di atasnya.

### 2.1.6 Match-Case (Pencocokan Pola Deklaratif)

Hadir secara eksklusif mulai Python 3.10 (merujuk pada PEP 634 dan 635), match-case menawarkan pendekatan yang jauh lebih canggih daripada switch-case tradisional. Bukannya sekadar menguji kesamaan sebuah nilai secara kaku, fitur ini bekerja sebagai instrumen pencocokan pola (structural pattern matching) yang sangat fleksibel. Match-case mampu membongkar komponen dari tipe data majemuk (seperti list atau dictionary) dan menarik informasi darinya dalam satu langkah. Alhasil, fitur ini menjadi alternatif penulisan kode yang jauh lebih rapi, terstruktur, dan elegan untuk menggantikan tumpukan IF-ELIF yang terlalu panjang.

### 2.2.1 Program Nested IF "A_2611532007"

```python
# Buat file dengan nama if_elif_else1_NIM.py
# Buat program untuk kondisional if 
# Nama variabel ditambah 4 digit nim terakhir, contoh: ipk_1234
# Program ini menggunakan fungsi input ()

umur_2007 = int(input("Input umur anda: "))
sim_2007 = (input("Apakah Anda Sudah Punya Sim C (y/t)")) [0]

if umur_2007 >= 17 and sim_2007 == 'y': 
    print("Anda Sudah deawsa dan bole bawa motor")

elif umur_2007 >= 17 and sim_2007 != 'y':
    print("Anda Sudah deawasa tetapi tidak boleh membawa motor")

elif umur_2007 < 17 and sim_2007 == 'y':
    print("Anda Belum Cukup Umur punya SIM")

else :
    print("Anda Belum Cukup Umur bawa motor")

print("Program Selesai")  
```

### 2.2.4 Program Match Castel "A_2611532007"

```python
# Buat file dengan nama match_castel_NIM.py
# Buat program untuk match castel
# Nama Variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input ()
# Program konversi nilai angka menjadi nama bulan

bulan_2007 = int(input("Masukkan angka bulan (1 - 12)=  "))

match bulan_2007:
    case 1:
        print("Januari")
    case 2 :
        print("Februari")
    case 3 :
        print("Maret")
    case 4 :
        print("April")
    case 5 :
        print("Mei")
    case 6 :
        print("Juni")
    case 7:
        print("juli")
    case 8 :
        print("Agustus")
    case 9 :
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12 :
        print("Desember")
    case _:
        print("Angka tidak valid")
```

Berikut penjelasan program :

1. Baris 1,2,3,4, dan 5 adalah komentar (comment)
2. Baris 7 program meminta pengguna memasukkan angka (1 - 12), lalu fungsi int() mengubah teks masukan tersebut menjadi angka bilangan bulat, dan menyimpannya di dalam variabel bernama bulan_1007.
3. Baris 9 program memulai blok percabangan match-case untuk mencocokkan nilai yang ada di dalam variabel bulan_1007 dengan berbagai kemungkinan yang ada. 
4. Baris 10-33 program mencocokkan nilai input secara berurutan; jika nilainya sama dengan 1, program akan mencetak "Januari", jika nilainya 2 mencetak "Februari", dan seterusnya hingga case 12 untuk "Desember"
5. Baris 34 simbol underscore (_) ini berfungsi sebagai jalan keluar terakhir (default case) yang akan dieksekusi jika angka yang dimasukkan oleh pengguna tidak cocok dengan seluruh nilai case di atasnya (misalnya pengguna memasukkan angka 13 atau 0).
6. Baris 35 jika program masuk ke kondisi case _:, maka program akan mencetak teks "Angka tidak valid" untuk memberitahu bahwa masukan berada di luar rentang 1 - 12.

### 2.2.6 Program Multi IF 2 "A_2611532007"

```python
# Buat file dengan nama multi_if2_NIM.py
# Buat program untuk kondisional if 
# Nama variabel ditambah 4 digit nim terakhir, contoh: total_belanja_1234
# Program ini menggunakan fungsi input ()
# Program ini menghitung diskon belanja

# Input dari user
total_belanja_2007 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2007 = input ("Apakah Anda Member? (y/t):").strip().lower
is_member_2007 = input_member_2007 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2007 = input("Apakah kode promo calid? (y/t):").strip().lower
kode_promo_valid_2007 = input_promo_2007 in ["y", "ya"]

total_diskon_persen_2007 = 0

# Multi if terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus
if total_belanja_2007 > 1000000:
    total_diskon_persen_2007 += 10 # Diskon belanjan besar

if is_member_2007:
    total_diskon_persen_2007 += 5 # Diskon member

if kode_promo_valid_2007:
    total_diskon_persen_2007 += 15 # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_2007 = total_belanja_2007 * (total_diskon_persen_2007 / 100)
total_bayar_2007 = total_belanja_2007 - nominal_diskon_2007

# Output Hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_2007}% (Rp {nominal_diskon_2007:,.0f})")
print(f"Total Bayar   : Rp {total_bayar_2007:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_2007}%")
# Output: Total diskon yang Anda dapatkan : 30% jika belanja > 1 juta , member, kode promo valid
```

Berikut penjelasan program :

1. Baris 1,2,3,4, dan 5 adalah kometar (comment)
2. Baris 8 program meminta pengguna memasukkan nominal total belanja, mengubahnya menjadi angka desimal (float), dan menyimpannya di variabel total_belanja_1007.
3. Baris 11-12 program meminta status member, membersihkan spasi berlebih (strip()), dan mengecek apakah teks yang dimasukkan adalah "y" atau "ya", lalu menyimpannya sebagai nilai True / False di variabel is_member_1007.
4. Baris 15-16 program mengecek apakah pengguna memiliki kode promo yang valid (menjawab "y" atau "ya") dan menyimpan status kebenarannya di variabel kode_promo_valid_1007.
5. Baris 18 program menyiapkan wadah penampung persentase diskon awal, yaitu variabel total_diskon_persen_1007 yang diberi nilai 0.
6. Baris 21-22 pengecekan kondisi pertama; jika total belanja lebih besar dari Rp 1.000.000, maka program akan menambahkan diskon sebesar 10% ke dalam total persentase.

Setelah data terkumpul, sistem melakukan satu pengecekan sederhana (IF Tunggal) terhadap jumlah tiket. Jika pengunjung memasukkan angka nol atau negatif pada jumlah tiket, sistem hanya akan memunculkan satu pesan peringatan tanpa menghentikan program, lalu lanjut ke tahap berikutnya.

2. Pemilihan Paket Wahana (Sistem Match-Case)
Sistem kemudian menampilkan daftar lima wahana beserta harganya. Pengunjung diminta memilih dengan mengetikkan angka satu sampai lima. Di sini, sistem menggunakan metode pencocokan pola. Jika angka yang dimasukkan cocok dengan salah satu dari lima pilihan tersebut, sistem akan menetapkan harga yang sesuai. Namun, jika pengunjung memasukkan angka yang tidak terdaftar, sistem akan masuk ke kondisi *default* (kasus lain-lain) yang menyatakan bahwa pilihan tidak valid.

3. Uji Kelayakan Pengendara (Sistem IF-ELIF-ELSE)
Tahap ini sangat ketat karena melibatkan keselamatan, khususnya untuk wahana Motor ATV Ekstrim. Sistem mengevaluasi syarat berlapis secara berurutan:

* Jika pengunjung memilih ATV, sudah dewasa, dan punya SIM, maka diizinkan membawa ATV sendiri.
* Jika memilih ATV, sudah dewasa, tapi tidak punya SIM, maka wajib didampingi instruktur.
* Jika memilih ATV, tapi masih di bawah umur, sistem akan langsung menolak, terlepas dari apa pun status SIM-nya (bahkan menganggap datanya tidak valid jika mengaku punya SIM).

4. Akumulasi Diskon Belanja (Sistem Multi IF)
Berbeda dengan pengecekan wahana yang saling menggugurkan, sistem perhitungan diskon ini bersifat mandiri dan bisa ditumpuk. Sistem akan mengecek empat syarat promosi satu per satu:

* Apakah total belanja mencapai dua ratus ribu rupiah? (Jika ya, tambah diskon 10%).
* Apakah pengunjung adalah seorang member? (Jika ya, tambah diskon 5%).
* Apakah kode promo yang dimasukkan terdaftar di sistem? (Jika ya, tambah diskon 15%).
* Apakah tiket yang dibeli berjumlah lima atau lebih? (Jika ya, tambah diskon 5%).
Karena sistemnya mengecek secara independen, pengunjung yang memenuhi semua syarat di atas bisa mendapatkan total tumpukan diskon hingga maksimal 35%.

5. Pembayaran dan Penentuan Bonus (Sistem IF-ELSE)
Setelah menghitung total harga yang sudah dipotong diskon, sistem akan mencetak rincian tagihan belanja. Di bagian paling akhir, sistem mengambil keputusan dua arah yang pasti. Jika total uang yang dibayarkan oleh pengunjung lebih dari tiga ratus ribu rupiah, sistem akan memberikan ucapan selamat sekaligus bonus suvenir. Sebaliknya, jika tidak mencapai nominal tersebut, sistem tetap merespons dengan memberikan ucapan terima kasih standar.

# <h1 align = "center">KESIMPULAN <h1> 

Berdasarkan praktikum yang telah dilaksanakan, dapat disimpulkan bahwa struktur kontrol percabangan (*selection structure*) memegang peranan fundamental dalam mengubah alur eksekusi program dari yang bersifat linear menjadi adaptif dan dinamis. Fitur ini memberikan kemampuan bagi sistem komputasi untuk mengevaluasi keadaan dan secara otomatis mengambil rute keputusan yang berbeda berdasarkan masukan atau kondisi data pada saat itu.

Dalam penerapannya, setiap varian struktur percabangan terbukti memiliki karakteristik dan tujuan logis yang sangat spesifik. Struktur *IF Tunggal* bertindak sebagai filter validasi satu arah, sementara *IF-ELSE* memastikan program selalu merespons dengan dua opsi tindakan yang saling meniadakan. Pada kasus yang lebih kompleks, *Nested IF* digunakan untuk mengunci evaluasi dengan prasyarat hierarkis yang berlapis. Perbedaan krusial juga terlihat antara *Multi IF* yang mengevaluasi banyak syarat secara mandiri dan akumulatif, berbanding terbalik dengan *IF-ELIF-ELSE* yang dirancang untuk rantai kondisi eksklusif di mana pengecekan langsung dihentikan saat kecocokan pertama ditemukan. Di sisi lain, penerapan *Match-Case* memberikan alternatif penulisan yang jauh lebih bersih dan terstruktur untuk skenario pencocokan pola yang banyak, menggantikan rantai keputusan yang terlalu panjang.

Secara keseluruhan, tantangan utama dalam merancang program percabangan bukanlah sekadar pada penulisan tata bahasa kode (*syntax*), melainkan pada ketajaman merumuskan kerangka logika. Kesalahan dalam memilih struktur yang tepat—seperti tertukarnya penggunaan logika akumulatif dengan logika yang saling menggugurkan—dapat merusak fungsionalitas dan kebenaran hasil akhir program. Oleh karena itu, pemahaman mendalam terhadap sifat masing-masing varian percabangan menjadi syarat mutlak untuk menghasilkan kode yang tidak hanya benar secara algoritma, tetapi juga efisien dan mudah dipahami (*readable*).
