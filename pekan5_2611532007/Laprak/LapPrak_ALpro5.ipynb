## 2.1 Dasar teori

Perulangan dalam bahasa  pemrograman berfungsi menyuruh komputer melakukan sesuatu secara berulang-ulang. Terdapat dua jenis perulangan dalam bahasa pemrograman python, yaitu perulangan dengan for dan while.

Perulangan for disebut counted loop (perulangan yang terhitung), sementara perulangan while disebut uncounted loop (perulangan yang tak terhitung). Perbedaannya adalah perulangan for biasanya digunakan untuk mengulangi kode yang sudah diketahui banyak perulangannya. Sementara while untuk perulangan yang memiliki syarat dan tidak tentu berapa banyak perulangannya

### 2.1.1 Perulangan For

#### 2.1.1.1 Pengertian
Mekanisme perulangan (looping) memungkinkan sebuah program untuk mengeksekusi serangkaian instruksi yang sama berkali-kali secara otomatis. Dalam bahasa Python, perulangan for memiliki karakteristik yang berbeda dibandingkan bahasa pemrograman tradisional (seperti C atau Java). Alih-alih menggunakan perhitungan variabel numerik untuk menentukan jumlah putaran, for loop pada Python dirancang untuk mengiterasi atau menelusuri elemen-elemen secara langsung dari sebuah objek iterable, seperti list, string, tuple, hingga dictionary.

#### 2.1.1.2 Sturktur Penulisan
Aturan penulisan perulangan di Python sangat intuitif karena menyerupai logika tata bahasa manusia. Format utamanya menggunakan konstruksi for [variabel_penampung] in [objek_data]:. Variabel penampung tersebut akan secara bergantian mewakili setiap nilai di dalam kumpulan data seiring berjalannya iterasi. Hal krusial lainnya di Python adalah penggunaan indentasi (spasi menjorok) untuk mendefinisikan batas blok kode mana yang akan dikenai perulangan.

#### 2.1.1.3 Fungsi Bawaan 'range()'
Untuk skenario di mana kita tidak memiliki sekumpulan data melainkan hanya ingin mengulang proses sebanyak $N$ kali, Python memfasilitasinya dengan fungsi range(). Fungsi ini bertugas menggenerasi urutan bilangan secara otomatis. Parameter fungsi ini terdiri dari tiga bagian, yakni: batas bawah (start), batas atas yang tidak diikutkan dalam iterasi (stop), serta jarak atau interval antar bilangannya (step).

#### 2.1.1.4 Pernyataan kendali
Dalam kondisi tertentu, alur perulangan yang berjalan normal dapat diinterupsi atau dimodifikasi menggunakan tiga perintah kendali utama:<br>
Break: Berfungsi untuk memberhentikan iterasi sepenuhnya secara paksa, sehingga program akan langsung keluar dari blok perulangan.<br>
Continue: Digunakan untuk mengabaikan atau melompati sisa kode pada putaran yang sedang berjalan, dan program akan langsung meloncat ke elemen iterasi berikutnya.<br>
Pass: Sebuah statement kosong yang tidak memberikan aksi apa pun. Instruksi ini umumnya dimanfaatkan sebagai kerangka sementara (placeholder) agar kode tidak menghasilkan error ketika blok perulangan belum diisi instruksi.<br>

#### 2.1.1.5 Neasted for
Nested loop merujuk pada situasi di mana kita mendefinisikan sebuah perulangan di dalam blok perulangan lain. Dalam konteks pemrosesan data, logika ini sangat fundamental untuk mengekstrak struktur data berdimensi ganda (seperti membaca tabel yang terdiri dari baris dan kolom). Secara operasional, perulangan yang berada di bagian terdalam (inner) wajib diselesaikan seluruh iterasinya terlebih dahulu sebelum perulangan bagian luar (outer) maju ke indeks berikutnya.

```python
# Perulangan for dalam python

ulang_2007 = int(input("Masukkan jumlah perulangan: "))

for i in range(ulang_2007):
    print(f"Perulangan ke-{i}")
```

```python
# perulangan for dalam python

ulang_2007 = int(input("Masukkan jumlah perulangan: "))

jumlah_2007 = 0
for i in range(1, ulang_2007 + 1):
    print(i, end=" ")
    jumlah_2007 = jumlah_2007 + i
    
    if i < ulang_2007:
        print("+",end=" ")
    else:
        print("=",jumlah_2007,)

print("jumlah =", jumlah_2007)
```

Mencetak simbol = diikuti dengan hasil penjumlahan yang ada di dalam variabel jumlah_2007. Karena baris ini tidak menggunakan parameter end, secara otomatis setelah selesai mencetak, kursor akan berpindah ke baris baru (enter).

Baris 15: print("jumlah =", jumlah_2007)
Mencetak kesimpulan di baris baru berupa teks "jumlah =" yang diikuti dengan nilai paling akhir (total keseluruhan) dari variabel jumlah_2007.

### 2.2.4 Program "nested_for1_2611532007"

```python
# Perulangan for dalam python

batas_2007 = int(input("Masukkan nilai batas: "))
for line_2007 in range(1, batas_2007+1):
    for j_2007 in range(1, (-1 * line_2007 + batas_2007) + 1):
        print(".", end=" ")
    print(line_2007)
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">
Penjelasan:

Baris 3: `batas_2007 = int(input("Masukkan nilai batas: "))`
Program meminta pengguna untuk memasukkan angka batas, mengubah input tersebut menjadi tipe data bilangan bulat (`int`), lalu menyimpannya ke dalam variabel bernama `batas_2007`.

Baris 4: `for line_2007 in range(1, batas_2007+1):`
Memulai perulangan luar (*outer loop*) dengan variabel `line_2007` yang nilainya berjalan dari angka 1 hingga nilai `batas_2007`. Penambahan `+1` dilakukan agar nilai batas yang diinputkan pengguna ikut diproses, karena fungsi `range` memotong angka batas akhirnya (eksklusif). Ini berfungsi untuk mengatur jumlah baris yang akan dicetak.

Baris 5: `for j_2007 in range(1, (-1 * line_2007 + batas_2007) + 1):`
Memulai perulangan dalam (*inner loop*) dengan variabel `j_2007` untuk mencetak karakter titik sebelum angka. Jumlah perulangan ini dihitung menggunakan rumus `(-1 * line_2007 + batas_2007) + 1`. Hasil rumus ini membuat jumlah perulangan semakin berkurang pada setiap barisnya, membentuk pola segitiga titik yang terbalik.

Baris 6: `print(".", end=" ")`
Mencetak karakter titik `.` untuk setiap perulangan dalam. Parameter `end=" "` digunakan agar titik-titik tersebut dicetak menyamping dalam satu baris yang sama dan dipisahkan oleh spasi, tidak pindah ke baris bawah.

Baris 7: `print(line_2007)`
Mencetak nilai angka dari variabel `line_2007` (nomor baris saat ini) pada akhir kumpulan titik tersebut. Karena tidak menggunakan parameter `end`, fungsi `print` ini secara otomatis akan memindahkan kursor ke baris baru (*enter*). Dengan begitu, perulangan `line_2007` selanjutnya akan mulai dicetak di baris bawahnya.

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">

Memulai perulangan luar (outer loop) dengan variabel i. Karena hanya menggunakan satu parameter di dalam fungsi range, perulangan ini secara otomatis dimulai dari angka 0. Penambahan + 1 dilakukan agar angka inputan terakhir (nilai batas_2007) ikut dieksekusi, karena batas akhir fungsi range bersifat eksklusif (tidak diikutkan).

Baris 5: for j in range(batas_2007+1):
Memulai perulangan dalam (inner loop) dengan variabel j. Jumlah iterasi pada perulangan dalam ini sama dengan perulangan luar, yaitu berjalan mulai dari 0 hingga nilai batas_2007. Ini berarti setiap satu kali perulangan i berjalan, perulangan j akan diselesaikan secara penuh terlebih dahulu.

Baris 6: print(i+j, end=" ")
Menjumlahkan nilai i dan nilai j pada perulangan saat ini, kemudian mencetak hasil penjumlahannya ke layar. Parameter end=" " digunakan agar hasil cetakan tersebut tidak memindahkan kursor ke baris baru, melainkan dilanjutkan menyamping dan dipisahkan oleh sebuah spasi.

Baris 7: print()
Mencetak baris kosong (enter). Perintah ini berada di luar perulangan dalam dan merupakan bagian dari perulangan luar. Fungsinya adalah untuk memutus cetakan menyamping dari baris ke-6 tadi, sehingga ketika nilai i bertambah dan masuk ke perulangan selanjutnya, hasil penjumlahan yang baru akan dicetak di baris yang baru.


### 2.2.7 Program nested_for4_2611532007
</div>

```python
tinggi_2007 = int(input("Masukkan tinggi pola (Bilangan genap, misal 10): "))

if tinggi_2007 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2007 = tinggi_2007
    c_2007 = a_2007
    lebar_2007 = (2 * tinggi_2007) - 2

    for i in range(1, tinggi_2007 + 1):
        b_2007 = c_2007 + 1
      
        for j in range(1, lebar_2007 + 1):
            # Baris atas dan bawah
            if i == 1 or i == tinggi_2007:
                if j == 1 or j == lebar_2007:
                    print("#", end="")
                else:
                    print("=", end="")
            else:
                # Sisi kiri dan kanan (border)
                if j == 1 or j == lebar_2007:
                    print("|", end="")
                else:
                    # Logika isi pola (diamond/kristal)
                    if j == c_2007:
                        print("<", end="")
                    elif j == b_2007:
                        print(">", end="")
                    elif j == (lebar_2007 - c_2007):
                        print("<", end="")
                    elif j == (lebar_2007 - c_2007 + 1):
                        print(">", end="")
                    elif j > b_2007 and j < (lebar_2007 - c_2007):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print() # Pindah ke baris baru

        # Logika asli java (Perubahan nilai a untuk membentuk pola)
        a_2007 -= 2  # KUNCI YANG HILANG: a harus dikurangi agar pola bergerak
        
        if a_2007 <= 0:
            c_2007 = (-a_2007) + 2
        else:
            c_2007 = a_2007
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">

Memulai perulangan luar (outer loop) yang berfungsi mengatur jumlah baris. Variabel i_2007 akan berjalan dari 1 hingga ke nilai tinggi_2007. Penambahan +1 diperlukan karena batas akhir pada fungsi range bersifat eksklusif (tidak ikut dieksekusi).
