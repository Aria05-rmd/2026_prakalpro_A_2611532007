## 2.1 Dasar Teori
Pengenalan tentang operator dalam python


1.  Operator Aritmatika<br>
Merupakan simbol-simbol yang difungsikan untuk menjalankan perhitungan matematis. Operator ini memfasilitasi berbagai kalkulasi dasar angka, mulai dari penambahan dan pengurangan, hingga operasi yang lebih spesifik seperti perkalian dan pembagian.

2.  Operator Penugasan (Assignment)<br>
Berfungsi untuk mengalokasikan atau memasukkan sebuah nilai ke dalam suatu variabel. Selain itu, operator ini juga berperan penting dalam memperbarui atau memodifikasi data yang sebelumnya telah tersimpan di dalam memori variabel tersebut.

3.  Operator Bitwise<br>
Digunakan untuk memanipulasi data pada level bit yang paling rendah. Proses komputasinya dilakukan dengan mengonversi nilai menjadi sistem bilangan biner (yang hanya terdiri dari digit 0 dan 1) sebelum kalkulasi dieksekusi oleh mesin.

4.  Operator Keanggotaan (Membership)<br>
Bertujuan untuk memvalidasi eksistensi suatu elemen tertentu. Operator ini akan mengevaluasi apakah sebuah nilai atau variabel merupakan bagian dari sebuah koleksi atau struktur data (seperti array, list, atau string).

5.  Operator Identitas (Identity)<br>
Bekerja dengan cara membandingkan lokasi memori dari dua objek. Operator ini bertugas untuk memastikan apakah dua variabel yang sedang dievaluasi sebenarnya merujuk pada entitas dan alamat memori fisik yang persis sama.

6.  Operator Perbandingan (Relasional)<br>
Berfungsi untuk mengkomparasi dua operan atau lebih. Keluaran yang dihasilkan dari operator ini selalu bersifat logis atau boolean (Benar/True atau Salah/False), yang merepresentasikan keabsahan dari kondisi yang sedang diuji.

7.  Operator Logika<br>
Berperan dalam menghubungkan, menggabungkan, atau mengevaluasi beberapa pernyataan kondisional sekaligus. Operator ini merupakan komponen fundamental dalam merancang alur kontrol dan proses pengambilan keputusan (decision making) di dalam eksekusi program.

## 2.2 Kode Program

### 2.2.1 Program "Operator Aritmatika_2611533011"
</div>

```python

# buat program untuk operator aritmatika dalam python
# nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# program ini menggunakan fungsi input()
# nilai yang di masukkan akan di konversi menjadi tipe data integer

angka1_3011 = int(input("input angka-1:"))
angka2_3011 = int(input("input angka-2:"))

# PENJUMLAHAN
hasil_3011 = angka1_3011 + angka2_3011
print("\nOperator Penjumlahan")
print("Hasil =", hasil_3011)

# PENGURANGAN
hasil_3011 = angka1_3011 - angka2_3011
print("\nOperator Pengurangan")
print("Hasil =", hasil_3011)

# PERKALIAN
hasil_3011 = angka1_3011 * angka2_3011
print("\nOperator Perkalian")
print("Hasil =", hasil_3011)

# PEMBAGIAN
if angka2_3011 != 0:
    hasil_3011 = angka1_3011 / angka2_3011
    print("\nOperator Pembagian")
    print("Hasil =", hasil_3011)

    hasil_3011 = angka1_3011 // angka2_3011
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_3011)

    hasil_3011 = angka1_3011 % angka2_3011
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_3011)
else:
    print("angka kedua tidak boleh bernilai 0.")

# PANGKAT
hasil_3011 = angka1_3011 ** angka2_3011
print("\n Operator Pangkat")
print("Hasil=", hasil_3011)
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">

Penjelasan:
1. Penjumlahan<br>
Angka pertama yaitu 7 dijumlahkan dengan angka kedua yaitu 8. Hasilnya adalah 15.<br>
7 + 8 = 15
2. Pengurangan<br>
Angka pertama dikurangi dengan angka kedua. Karena angka pertama lebih kecil dari angka kedua, hasilnya menjadi -1.
7 - 8 = -1
3. Perkalian<br>
Angka 7 dikalikan dengan angka 8 sehingga menghasilkan 56.<br>
7 × 8 = 56
4. Pembagian<br>
Angka 7 dibagi dengan angka 8. Hasilnya berupa bilangan desimal, yaitu 0.875.<br>
7 ÷ 8 = 0.875
5. Pembagian Bulat<br>
Pada pembagian bulat, hasil pembagian 7 dengan 8 hanya mengambil bagian bilangan bulatnya. Karena hasil pembagian adalah 0.875, maka hasil pembagian bulatnya adalah 0.<br>
7 // 8 = 0
6. Sisa Bagi<br>
Operasi sisa bagi digunakan untuk mengetahui sisa dari pembagian 7 dengan 8. Karena 7 masih lebih kecil dari 8, maka seluruh angka 7 menjadi sisa pembagian.<br>
7 % 8 = 7<br>
Jadi, hasilnya adalah 7.
7. Pangkat<br>
Pada operasi pangkat, angka pertama yaitu 7 menjadi bilangan dasar dan angka kedua yaitu 8 menjadi pangkat.
7⁸ = 5.764.801<br>
Sehingga hasil yang ditampilkan adalah 5764801.

### 2.2.2 Program "Operator Assignment_2611533011"
</div>

```python
# nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# program ini menggunakan fungsi input()
# nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# program operator assignment dalam python

angka1_3011 = int(input("input angka-1: "))
angka2_3011 = int(input("input angka-1: "))

print("\n nilai awal angka1 =", angka1_3011)
print("nilai angka2 =", angka2_3011)

# assignment biasa 
hasil_3011 = angka1_3011
print("\n assingmnet biasa (=)")
print("Hasil =", hasil_3011)

# assignment penambahan
hasil_3011 = angka1_3011
hasil_3011 += angka2_3011
print("\n assignment penambahan (+=)")
print("Hasil =", hasil_3011)

# assignment perkalian
hasil_3011 = angka1_3011
hasil_3011 *= angka2_3011
print("\n assignment perkalian (*=)")
print("Hasil =", hasil_3011)

# assignment pembagian, pembagian bulat, dan sisa bagi 
if angka2_3011 != 0:
    hasil_3011 = angka1_3011
    hasil_3011 /= angka2_3011
    print("\n assignment pembagian (/=)")
    print("Hasil =", hasil_3011)

    #operator tambahan
    hasil_3011 = angka1_3011
    hasil_3011 //= angka2_3011
    print("\n assignment pembagian bulat (//=)")
    print("Hasil =", hasil_3011)

    hasil_3011 = angka1_3011
    hasil_3011 %= angka2_3011
    print("\n assignment sisa bagi (%=)")
    print("Hasil =", hasil_3011)

else:
    print("\n pembagian tidak dapat dilakukan")
    print("angka kedua tidak boleh 0")
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">
Penjelasan:

1. Nilai Awal<br>
Program menampilkan nilai awal dari kedua variabel. angka1 memiliki nilai 8, sedangkan angka2 memiliki nilai 5.<br>
2. Assignment Biasa (`=`)<br>
Assignment biasa digunakan untuk memberikan nilai dari angka1 kepada variabel hasil. Karena nilai angka1 adalah 8, maka hasil yang ditampilkan adalah 8.<br>
Hasil = 8
3. Assignment Penambahan (`+=`)<br>
Operator += digunakan untuk menambahkan nilai angka2 ke nilai yang sudah terdapat pada hasil.
8 + 5 = 13<br>
Jadi, hasil yang ditampilkan adalah 13.
4. Assignment Perkalian (`*=`)<br>
Operator *= digunakan untuk mengalikan nilai hasil dengan nilai angka2.
8 × 5 = 40<br>
Jadi, hasil yang ditampilkan adalah 40.
5. Assignment Pembagian (`/=`)<br>
Operator /= digunakan untuk membagi nilai hasil dengan angka2.
8 ÷ 5 = 1.6<br>
Karena pembagian biasa dapat menghasilkan bilangan desimal, maka output yang diperoleh adalah 1.6.
6. Assignment Pembagian Bulat (`//=`)<br>
Operator //= digunakan untuk melakukan pembagian dan mengambil hasil bilangan bulatnya.
8 ÷ 5 = 1.6<br>
Bagian bilangan bulat dari 1.6 adalah 1, sehingga hasil yang ditampilkan adalah 1.
7. Assignment Sisa Bagi (`%=`)<br>
Operator %= digunakan untuk mendapatkan sisa dari pembagian antara angka1 dan angka2.
8 ÷ 5 = 1 sisa 3<br>
Jadi, hasil sisa bagi yang ditampilkan adalah 3.
### 2.2.3 Program "Operator Bitwise_2611533011"
</div>

```python
# nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# program ini menggunakan fungsi input()


print("\n========================================")
print("3. Operator Bitwise")
print("==========================================")

angka1_3011 = int(input("Masukkan angka bitwise-1: "))
angka2_3011 = int(input("masukkan angka bitwise-2: "))

print("\n angka dalam bentuk desimal dan biner")
print("angka1 =", angka1_3011, "| biner = ", bin(angka1_3011))
print("angka2 =", angka2_3011, "| biner = ", bin(angka2_3011))

# Bitwise AND
hasil_3011 = angka1_3011 & angka2_3011
print("\n Bitwise AND (&)")
print(angka1_3011, "&", angka2_3011, "=", hasil_3011)
print("Biner Hasil =", bin (hasil_3011))
print("Biner Hasil (8 bit) =", format(hasil_3011, "08b"))

# Bitwise OR
hasil_3011 = angka1_3011 | angka2_3011
print("\n Bitwise OR (|)")
print(angka1_3011, "|", angka2_3011, "=", hasil_3011)
print("Biner Hasil =", bin(hasil_3011))
print("Biner Hasil (8 bit) =", format(hasil_3011, "08b"))

# Bitwise XOR
hasil_3011 = angka1_3011 ^ angka2_3011
print("\n Bitwise XOR (^)")
print(angka1_3011, "^", angka2_3011, "=", hasil_3011)
print("Biner Hasil", bin(hasil_3011))
print("Biner Hasil (8 bit) =", format(hasil_3011, "08b"))

# Bitwise NOT
hasil_3011 = ~angka1_3011
print("\n Bitwise NOT (~)")
print("~", angka1_3011, "=", hasil_3011)
print("Biner Hasil", bin(hasil_3011))
print("Biner Hasil (8 bit) =", format(hasil_3011, "08b"))

# Bitwise geser kiri
jumlah_geser = int(input("\n masukkan jumlah pergeseran bit: "))

hasil_3011 = angka1_3011 << jumlah_geser
print("\n Bitwise geser kiri (<<)")
print(angka1_3011, "<<", jumlah_geser, "=", hasil_3011)
print("Biner Hasil", bin(hasil_3011))
print("Biner Hasil (8 bit) =", format(hasil_3011, "08b"))

# Bitwise geser kanan
hasil_3011 = angka1_3011 >> jumlah_geser
print("\n Bitwise geser kanan (>>)")
print(angka1_3011, ">>", jumlah_geser, "=", hasil_3011)
print("Biner Hasil", bin(hasil_3011))
print("Biner Hasil (8 bit) =", format(hasil_3011, "08b"))
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">
Penjelasan:

1. Bentuk Desimal dan Biner<br>
Angka pertama yang digunakan adalah 4, sedangkan angka kedua adalah 8.<br>
Dalam bentuk biner:<br>
4 = 100<br>
8 = 1000<br>
Bentuk biner ini digunakan karena operator bitwise bekerja pada setiap bit dari bilangan tersebut.<br>
2. Bitwise AND (&)<br>
Operasi AND membandingkan setiap bit dari kedua angka. Hasilnya bernilai 1 hanya jika kedua bit yang dibandingkan sama-sama 1.<br>
  0100<br>
& 1000<br>
  0000<br>
Sehingga:<br>
4 & 8 = 0<br>
Hasil dalam 8 bit adalah 00000000.<br>
3. Bitwise OR (|)<br>
Operasi OR menghasilkan nilai 1 apabila salah satu atau kedua bit bernilai 1.<br>
  0100<br>
| 1000<br>
  1100<br>
Biner 1100 memiliki nilai desimal 12.<br>
Sehingga:<br>
4 | 8 = 12<br>
Hasil dalam 8 bit adalah 00001100.<br>
4. Bitwise XOR (^)<br>
Operasi XOR menghasilkan nilai 1 apabila kedua bit berbeda.<br>
  0100<br>
^ 1000<br>
  1100<br>
Karena posisi bit 1 pada angka 4 dan angka 8 berbeda, hasilnya adalah 1100, yaitu 12 dalam desimal.<br>
Sehingga:<br>
4 ^ 8 = 12<br>
Hasil dalam 8 bit adalah 00001100.<br>
5. Bitwise NOT (~)<br>
Operator NOT digunakan untuk membalik bit. Pada Python, operasi NOT menggunakan representasi bilangan bertanda sehingga hasil dari ~4 adalah -5.<br>
Secara matematis, operasi NOT pada bilangan 4 dapat dipahami dengan rumus:<br>
~n = -(n + 1)<br>
Maka:<br>
~4 = -(4 + 1) = -5<br>
Oleh karena hasilnya negatif, tampilan biner dengan format "08b" juga akan menampilkan tanda negatif, sehingga outputnya adalah:<br>
-00000101<br>
6. Bitwise Geser Kiri (<<)<br>
Operator geser kiri digunakan untuk menggeser bit ke sebelah kiri sebanyak jumlah pergeseran yang ditentukan.<br>
Jumlah pergeseran adalah 3.<br>
4 = 00000100<br>
     ↓ geser kiri 3 bit<br>
   00100000<br>
Hasilnya adalah 32.<br>
Sehingga:<br>
4 << 3 = 32<br>
Setiap pergeseran ke kiri satu posisi secara umum setara dengan mengalikan bilangan dengan 2. Karena digeser 3 kali:<br>
4 × 2³ = 4 × 8 = 32<br>
Hasil biner 8 bit adalah 00100000.<br>
7. Bitwise Geser Kanan (>>)<br>
Operator geser kanan digunakan untuk menggeser bit ke sebelah kanan sebanyak 3 posisi.<br>
4 = 00000100<br>
     ↓ geser kanan 3 bit<br>
   00000000<br>
Karena angka 4 digeser ke kanan sebanyak 3 posisi, semua bit 1 akan keluar dari posisi yang digunakan.<br>
Sehingga:<br>
4 >> 3 = 0<br>
Hasil biner 8 bit adalah 00000000.<br>

### 2.2.4 Program Operator Keanggotaan_2611533011

```python
print("=========================")
print(" 1. OPERATOR KEANGGOTAAN")
print("=========================")

# input beberapa data yang dipisahkan dengan koma
input_data_3011 = input("masukkan beberapa angka, pisahkan dengan koma: ")

# mengubah input menjadi list integer
data_3011 = [int(angka.strip()) for angka in input_data_3011.split(",")]

nilai_dicari_3011 = int(input("masukkan angka yang ingin di cari: "))

#operator in 
hasil_3011 = nilai_dicari_3011 in data_3011 
print("\n Operator keanggotaan IN")
print(nilai_dicari_3011, "in", data_3011, "=", hasil_3011)

# operator not in 
hasil_3011 = nilai_dicari_3011 not in data_3011 
print("\n Operator keanggotaan NOT IN")
print(nilai_dicari_3011, "not in", data_3011, "=", hasil_3011)
```

Karena nilai 9 memang tidak terdapat dalam list [8, 7], maka kondisi not in bernilai:<br>
9 not in [8, 7] = True<br>
Artinya, angka 9 benar-benar tidak terdapat dalam data [8, 7].<br>

### 2.2.5 Program Operator Identitas_2611533011
</div>

```python
print("\n======================")
print("2. OPERATOR IDENTITAS")
print("========================")

# objek1 menggunakan list dari input pengguna 
objek1_3011 = data_3011

# objek2 merujuk pada objek yang sama dengan objek1
objek2_3011 = objek1_3011 

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_3011 = data_3011.copy()

print("objek1_3011 =", objek1_3011)
print("objek2_3011 =", objek2_3011)
print("objek3_3011 =",objek3_3011)

# operator is
hasil_3011= objek1_3011 is objek2_3011
print("\n Operator identitas IS")
print("objek1_3011 is objek2_3011 =", hasil_3011)

# operator is not
hasil_3011 = objek1_3011 is objek3_3011
print("\n Operator identitas IS NOT")
print("objek1_3011 is not objek3_3011 =", hasil_3011)

# membandingkan identitas dan nilai 
print("\n membandingkan identitas dan nilai")
print("objek1_3011 is objek3_3011: ", objek1_3011 is objek3_3011)
print("objek1_3011 == objek3_3011: ", objek1_3011 == objek3_3011)
```

4. Operator is<br>
Perbandingan objek1_3011 is objek2_3011 menghasilkan True karena kedua variabel merujuk pada objek yang sama.<br>
5. Operator is not<br>
Pada program terdapat perbandingan identitas antara objek1_3011 dan objek3_3011. Karena keduanya merupakan objek yang berbeda, maka kondisi is not bernilai True jika ditulis sesuai operator tersebut.<br>
Namun, pada kode yang diberikan, baris:<br>
objek1_3011 is not objek3_3011<br>
seharusnya menghasilkan True, bukan False. Jadi output yang benar untuk bagian tersebut adalah:<br>
objek1_3011 is not objek3_3011 = True<br>
6. Perbandingan Identitas<br>
Perbandingan objek1_3011 is objek3_3011 menghasilkan False karena kedua objek memiliki identitas yang berbeda, meskipun isinya sama.<br>
7. Perbandingan Nilai<br>
Perbandingan objek1_3011 == objek3_3011 menghasilkan True karena isi kedua objek sama, yaitu [8, 7].<br>
8. Kesimpulan<br>
Berdasarkan hasil program, operator identitas digunakan untuk mengetahui apakah dua variabel mengacu pada objek yang sama. Operator is membandingkan identitas objek, sedangkan == membandingkan nilai atau isi objek. Dalam percobaan ini, objek1_3011 dan objek2_3011 merupakan objek yang sama, sedangkan objek3_3011 merupakan objek baru yang memiliki isi sama.

```python
# nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# program operator perbandingan dalam python

angka1_3011 = int(input("input angka-1:"))
angka2_3011 = int(input("input angka-2:"))

# Lebih besar dari
hasil_3011 = angka1_3011 > angka2_3011
print("\n Operator lebih besar dari")
print("angka1 > angka 2 =", hasil_3011)

# Lebih kecil dari
hasil_3011 = angka1_3011 < angka2_3011
print("\n Operator lebih kecil dari")
print("angka1 < angka 2 =", hasil_3011)

# Lebih besar dari atau sama dengan
hasil_3011 = angka1_3011 >= angka2_3011
print("\n Operator lebih besar dari atau sama dengan")
print("angka1 >= angka 2 =", hasil_3011)

# Lebih kecil dari atau sama dengan
hasil_3011 = angka1_3011 <= angka2_3011
print("\n Operator lebih kecil dari atau sama dengan")
print("angka1 <= angka 2 =", hasil_3011)

#  Sama dengan
hasil_3011 = angka1_3011 == angka2_3011
print("\n Operator sama dengan")
print("angka1 == angka 2 =", hasil_3011)

# tidak sama dengan
hasil_3011 = angka1_3011 != angka2_3011
print("\n Operator tidak sama dengan")
print("angka1 != angka 2 =", hasil_3011)

# Tambahan: perbandingan berarti dalam python
hasil_3011 = 0 < angka1_3011 < 100
print("\n Perbandingan berantai")
print("0 < angka1 < 100 =", hasil_3011)

hasil_3011 = 0 < angka2_3011 < 100
print("0 < angka2 < 100 =", hasil_3011)
```

Program membandingkan apakah angka1 lebih kecil daripada angka2.<br>
6 < 9 = True<br>
Hasilnya True karena angka 6 memang lebih kecil daripada angka 9.

Operator ini digunakan untuk memeriksa apakah angka1 lebih besar atau sama dengan angka2.<br>
6 >= 9 = False<br>
Hasilnya False karena angka 6 tidak lebih besar dan juga tidak sama dengan angka 9.
4. Operator Lebih Kecil Dari atau Sama Dengan (<=)<br>
Operator ini digunakan untuk memeriksa apakah angka1 lebih kecil atau sama dengan angka2.<br>
6 <= 9 = True<br>
Hasilnya True karena angka 6 lebih kecil daripada angka 9.
5. Operator Sama Dengan (==)<br>
Operator ini digunakan untuk memeriksa apakah angka1 dan angka2 memiliki nilai yang sama.<br>
6 == 9 = False<br>

6. Operator Tidak Sama Dengan (!=)<br>
Operator ini digunakan untuk memeriksa apakah angka1 dan angka2 memiliki nilai yang berbeda.<br>
6 != 9 = True<br>
Hasilnya True karena angka 6 dan 9 memiliki nilai yang berbeda.<br>

0 < 6 < 100 = True<br>

### 2.2.7 Program Operator Logika_2611533011

```python
# nama variabel ditambah 4 digit nim terakhir contoh: a1_1234
# program ini menggunakan fungsi input()
# program operator logika dalam python

# Memasukkan nilai boolean
# input tidak peka terhadap huruf besar dan kecil
a1_3011 = input("input nilai booelan-1 (true/false): "). strip(). lower() == "true"
a2_3011 = input("input nilai booelan-2 (true/false): "). strip(). lower() == "true"

print("\nA1 =", a1_3011)
print("A2 =", a2_3011)

# konjungsi: bernilai True jika keduanya True
hasil_3011 = a1_3011 and a2_3011
print("\n konjungsi (AND)")
print("A1 and A2 =", hasil_3011)

# disjungsi: bernilai true jika salah satunya true
hasil_3011 = a1_3011 or a2_3011
print("\n disjungsi (OR)")
print("A1 or A2 = ", hasil_3011)

# negasi A1: membalik nilai A1
hasil_3011 = not a1_3011 
print("\n negasi (NOT)")
print("not A1= ", hasil_3011)

# negasi A2: membalik nilai A2
hasil_3011 = not a2_3011 
print("\n negasi (NOT)")
print("not A2= ", hasil_3011)

# XOR: bernilai true jika kedua nilai berbeda
hasil_3011 = a1_3011 != a2_3011
print("\n disjungsi eksklusif (XOR)")
print("A1 XOR A2 =", hasil_3011)
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">
Penjelasan:

1. Nilai A1 dan A2<br>
Program meminta input berupa nilai Boolean, yaitu true atau false. Karena input yang diberikan adalah 4 dan 7, keduanya tidak sama dengan teks "true". Oleh karena itu, program mengubah kedua nilai tersebut menjadi:<br>
A1 = False<br>
A2 = False<br>
2. Operator Konjungsi (AND)<br>
Operator and menghasilkan True jika kedua nilai bernilai True. Karena A1 dan A2 sama-sama False, maka:<br>
False and False = False<br>
Jadi, hasilnya adalah False.
3. Operator Disjungsi (OR)<br>
Operator or menghasilkan True jika salah satu atau kedua nilai bernilai True. Karena A1 dan A2 sama-sama False, maka:<br>
False or False = False<br>
Jadi, hasilnya adalah False.<br>
4. Negasi (NOT) A1<br>
Operator not digunakan untuk membalik nilai Boolean. Karena A1 bernilai False, maka:<br>
not False = True<br>
Jadi, hasil negasi A1 adalah True.
5. Negasi (NOT) A2<br>
Sama seperti A1, A2 juga memiliki nilai False. Setelah menggunakan operator not, nilainya menjadi:<br>
not False = True<br>
Jadi, hasil negasi A2 adalah True.
6. Operator XOR<br>
XOR menghasilkan True jika kedua nilai berbeda. Dalam percobaan ini A1 dan A2 sama-sama False, sehingga keduanya tidak berbeda.<br>
False XOR False = False<br>
Jadi, hasil XOR adalah False.

```python
# Program kasir sederhana

nama_2007 = str(input("Masukkan nama pelanggan  : "))
status_2007 = str(input("Masukkan status pelanggan (member/nonmember) : ")).strip().lower()
total_2007 = int(input("Masukkan total belanja : "))
jumlah_2007 = int(input("Masukkan jumlah barang : "))
kode_2007 = str(input("Masukkan kode promo : "))

kode_promo_tersedia_2007 = ["HEMAT10","HEMAT20","GRATISONGKIR"]

print("\n===DATA TRANSAKSI===") 
print("Nama Pelanggan          : ", nama_2007)
print("Status Pelanggan        : ", status_2007)
print("Jumlah Barang           : ", jumlah_2007)
print("Kode Promo              : ", kode_2007)

print("\n===HASIL VALIDASI===")

# Menggunakan Operator Perbandingan
v_belanja = total_2007 >= 200000
v_jumlah = jumlah_2007 >= 3
v_status = status_2007 == "member"

# Menggunakan Operator Keanggotaan
v_kode = kode_2007 in kode_promo_tersedia_2007

# Asumsi aturan diskon & promo
v_diskon = v_belanja
v_promo = v_jumlah

print("Belanja >= 200000       : ", v_belanja)
print("Jumlah Barang >= 3      : ", v_jumlah)
print("Status == Member        : ", v_status)
print("Kode Promo Tersedia     : ", v_kode)
# Menggunakan Operator Logika
print("Mendapatkan Diskon      : ", v_diskon or v_promo)
print("Mendapatkan Promo       : ", v_promo)

# Menghitung Diskon
if kode_2007 == "HEMAT10":
    diskon_2007 = total_2007 * 0.10
elif kode_2007 == "HEMAT20":
    diskon_2007 = total_2007 * 0.20
else: 
    diskon_2007 = 0

# Menggunakan Operator Assignment (-=)
total_2007 -= diskon_2007

print("\n=== HASIL PERHITUNGAN ===")
print("Diskon                  : ", diskon_2007)
print("Total Pembayaran        : ", total_2007 )
# Menggunakan Operator Aritmatika (/)
print("Rata - rata Harga Barang: ", total_2007 / jumlah_2007 if jumlah_2007 > 0 else 0)

print("\n=== HAK AKSES PELANGGAN ===")
# Contoh isi "....." (Bisa disesuaikan dengan logika Anda)
print("Kode Hak Akses          : ", id(nama_2007)) # Menggunakan id() sebagai contoh
print("Member Access           : ", v_status)
print("Promo Access            : ", v_promo)
print("Free Shiping Access     : ", kode_2007 == "GRATISONGKIR")

# --- Bagian Bitwise ---
# Menghapus deklarasi ganda yang ada di kode awal
kode_transaksi_1021 = int(v_status) << 0 | int(v_belanja) << 1 | int(v_jumlah) << 2 | int(v_promo) << 3
kode_referensi_1021 = int(v_status) << 0 | int(v_belanja) << 1 | int(v_promo) << 3

print("\n=== OPERASI BITWISE ===")
print("=== Kode Status Transaksi ===")
# PERBAIKAN: Menggunakan tanda kutip tunggal ('04b') di dalam f-string
print(f"{format(int(v_status) << 0, '04b')} | {format(int(v_belanja) << 1, '04b')} | {format(int(v_jumlah) << 2, '04b')} | {format(int(v_promo) << 3, '04b')}")
print(f"Kode Biner   : {format(kode_transaksi_1021, '04b')}")
print(f"Kode Desimal : {kode_transaksi_1021}")

print("\n=== Pemeriksaan Status ===")
print("Cek Member")
print(f"{format(kode_transaksi_1021, '04b')} & {format(int(v_status) << 0, '04b')}")
print(f"Hasil Biner   : {format((kode_transaksi_1021) & int(v_status) << 0, '04b')}")
print(f"Hasil Desimal : {(kode_transaksi_1021) & int(v_status) << 0}")

print("Cek Promo")
print(f"{format(kode_transaksi_1021, '04b')} & {format(int(v_promo) << 3, '04b')}")
print(f"Hasil Biner   : {format((kode_transaksi_1021) & int(v_promo) << 3, '04b')}")
print(f"Hasil Desimal : {(kode_transaksi_1021) & int(v_promo) << 3}")

print("\n=== Perbandingan Status ===")
print(f"Kode Transaksi : {format(kode_transaksi_1021, '04b')}")
print(f"Kode Referensi : {format(kode_referensi_1021, '04b')}")
print(f"{format(kode_transaksi_1021, '04b')} ^ {format(kode_referensi_1021, '04b')}")
print(f"Hasil Biner   : {format((kode_transaksi_1021) ^ (kode_referensi_1021), '04b')}")
print(f"Hasil Desimal : {(kode_transaksi_1021) ^ (kode_referensi_1021)}")

print("\n=== Shift ===")
print(f"{format(kode_transaksi_1021, '04b')} << 1")
print(f"Hasil Biner   : {format((kode_transaksi_1021) << 1, '04b')}")
print(f"Hasil Desimal : {(kode_transaksi_1021) << 1}") 
print("=== SELESAI ===")
```
