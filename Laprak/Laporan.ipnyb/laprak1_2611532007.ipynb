## 2.1 Dasar Teori

### 1. Python 
 Python merupakan bahasa pemrograman komputer yang biasa dipakai untuk membangun situs, software/aplikasi, mengotomatiskan tugas dan melakukan analisis data. Bahasa pemrograman ini termasuk bahasa tujuan umum. Artinya, ia bisa digunakan untuk membuat berbagai program berbeda, bukan khusus untuk masalah tertentu saja.
 Karena sifatnya yang serba guna dan mudah digunakan, ia menjadi bahasa pemrograman yang paling banyak digunakan. Terutama untuk mereka yang masih pemula.<br>
 Python tidak hanya mudah dipelajari tapi juga fleksibel. Ada lebih dari 125.000 library dari pihak ketiga yang memungkinkan kamu menggunakannya untuk machine learning, pemrosesan website bahkan biologi.<br>
 Selain itu, library yang berfokus pada data seperti pandas, NumPy dan matplotlib membuatnya sangat mudah memproses, memanipulasi serta memvisualisasikan data. Itulah sebabnya banyak orang yang suka menggunakannya dalam analisis data.


### 2. Visual Studio Code (VS Code)  
 Visual Code Studio merupakan code editor gratis yang dapat dijalankan pada perangkat desktop dengan basis Linux, Windows, dan Mac OS yang mana dikembangkan oleh Microsoft. Aplikasi visual studio code adalah sebuah software editor powerful namun tetap ringan saat digunakan.

 Visual Editor Code memiliki beberapa fitur yang menarik, berikut fitur fitur vs code.<br>

  a. Basic Editing<br>
     Sesuai namanya, fungsi visual studio code adalah coding ataupun membuat aplikasi sebagai fungsi dasarnya yang memang tidak perlu diragukan lagi. Fitur yang ada seperti column selection, multiple selection, dan keyboard shortcut.<br>
  b. IntelliSense<br>
     Fitur yang satu ini serupa dengan Autocomplete yang dapat membantu developer pada proses coding serta memberikan saran secara keseluruhan berdasarkan pada kata yang diketik. Fitur ini hanya tersedia pada beberapa bahasa pemrograman seperti CSS, HTML, Javascript, Typescript, dan juga JSON.<br>
  c. Debugging<br>
     fitur ini bisa membantu kalian dalam mengeksekusi, mengedit, dan juga meng-compile kode dengan berulang atau yang dikenal dengan sebutan looping<br>
  d. Extension Marketplace<br>
     Fitur ini tergolong paling penting kita bisa secara mudah mengunduh lalu memasangnya di berbagai extension yang diperlukan,bahkan bahasa pemrograman.<br>
  e. Github Integration<br>
     Fitur ini tergolong paling penting serta menjadi suatu fitur unggulan jika dibandingkan pesaingnya. Kalian bisa secara mudah mengunduh lalu memasangnya di berbagai extension yang diperlukan semacam debuggers dan tools.


### 3. Git dan GitHub 
Git adalah sistem kontrol versi terdistribusi yang memungkinkan pengembang untuk melacak perubahan yang dilakukan pada kode sumber selama pengembangan perangkat lunak,Dengan Git, setiap perubahan yang dilakukan pada kode disimpan dalam repositori, sehingga memudahkan pengembang untuk mengelola versi dan riwayat perubahan kode.<br>
GitHub adalah layanan berbasis cloud yang menyediakan tempat untuk meng-host repositori Git. Dengan GitHub, pengembang dapat menyimpan proyek mereka secara daring, berbagi kode, berkolaborasi dengan orang lain, dan mengelola versi perangkat lunak mereka secara lebih efisien<br>

Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 pratikum alpro
"""
print("Hello World")
print()
print("program ini menghasilkan")
print("empat baris output")
print("/ \\ // \\\ ///")
print("program ini" +" dibuat oleh a/n"
+ "Aria Ramadhan 2611532007" )
```

**Penjelasan :**
1. Baris 1-6 adalah baris komentar ini cuma catatan/keterangan buat programmer, bukan perintah.(#) untuk komentar 1 baris, sedangkan ("""...""") untuk komentar banyak baris.
