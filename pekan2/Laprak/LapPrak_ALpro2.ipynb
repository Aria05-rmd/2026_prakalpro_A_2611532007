## 2.1 Dasar Teori

### 2.1.1 Tipe Data dan Variable
#### 2.1.1.1 Definisi 

 Variable dalam python adalah nama yang digunakan untuk menetapkan sebuah nilai atau objek. Secara teknis variable adalah referensi atau label yang menunjuk ke suatu nilai dalam memori komputer. Variable digunakan untuk menyimpan data yang dapat diakses dan diubah selama program berjalan. Tipe data dalam Python adalah jenis nilai yang dapat ditampung oleh sebuah variabel untuk diproses oleh program.

 Dalam python,anda tidak perlu mendeklarasikan tipe data secara eksplisit seperti beberapa bahasa pemrograman lain.Python menggunakan konsep "Dynamic typing",yang berarti tipe data variable ditentukan secara otomatis sesuai nilai yang diberikan.

#### 2.1.1.2 Macam - macam Tipe Data dan Variable Dalam Python

Di Python,terdapat beberapa jenis variable yang bisa digunakan. Beberapa jenis variable yang umum di gunakan di python antara lain

1. Variable string<br>Variable String adalah variable yang digunakan untuk menyimpan teks. String dalam python dapat ditulis dalam tanda kutip tunggal(''), tanda kutip ganda("") dan tanda kutip tiga("'"' atau '"'"). Variable juga mendukung operasi penggambungan (concatenatian), pengindeksan (indexing) dan sclising. Variable ini bersifat imutable artinya setelah string dibuat tidak dapat dirubah tapi dapat membuat string baru dari string yang sudah ada.

2. Variable Interger<br>Variable Interger adalah variable yang diguakan untuk menyimpan bilangan bulat(tanpa desimal). Variable ini dapat dibuat dengan menetapkan nilai langsung atau  dengan operasi matematika. Variable integer dalam python
memiliki batas atas atas dan batas bawah, umumnya batas atasnya 2 miliar untuk integer signed (bilangan bulat dengan tanda) dan 4 miliar untuk integer unsigned (bilangan bulat tanpa tanda), namun batasnya berbeda tergantung versi python yang digunakan.

3. Variable float<br>Variable Float adalah variable yang digunakan untuk menyimpan bilangan pecahan atau desimal. Walaupun kemampuannya dalam menyimpan bilangan desimal dalam presisi tinggi, namun meiliki keterbatasan dalam reprentasi angka desimal tertentu yang menyebabkan beberapa operasi matematika menghasilkan nilai tidak tepat.

4. Boolean<br>Variable Boolean adalah variable yang memiliki 2 nilai: True atau False. Variable ini digunakan untuk menyimpan nilai kebenaran yang sering di gunakan dalam sistem alur program seperti kondisional(if,elif dan else) dan loop(while dan for).

Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Buat file dengan nama Boolean_2611532007.py
# Nama variable ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasi variable dengan tipe data Boolean
is_lulus_2007 = True
is_cumlaude_2007 = True

# Menggunakan Boolean
nilai_2007 = 85
batas_lulus_2007 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan_2007 = nilai_2007 >= batas_lulus_2007  # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai_2007)
print("Apakah Lulus?:", status_kelulusan_2007)
if is_lulus_2007 and is_cumlaude_2007 :
    print("Selamat, anda lulus dengan predikat Cumlaude")
```

5. Baris 9 Membuat variabel batas_lulus_2007 bernilai 75 (tipe integer).
7. Baris 12 Mengevaluasi apakah 85 >= 75. Karena nilainya benar, variabel status_kelulusan_2007 menyimpan nilai Boolean True.
8. Baris 14 Mencetak teks judul "=== Check Kelulusan ===" ke layar.

### 2.2.4 Input_2611532007
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Buat file dengan nama Input_2611532007.py
# Buat program untuk menghitung luas persegi panjang
# Nama variable ditambah 4 digit nim terakhir contoh: panjang_1234
# Program ini menggunakan fungsi input() untuk meminta pengguna 
# Memasukan nilai panjang dan lebar persegi panjang.
# Nilai yang dimasukkan akan dikonversi menjadi tipe data interger
panjang_2007 = input('masukkan nilai panjang: ')
lebar_2007 = input('masukkan nilai lebar: ')
luas_2007 = int(panjang_2007) * int(lebar_2007)
print("luas =", luas_2007)
```

4. Baris 9 Mengubah (*type casting*) isi variabel panjang_2007 dan lebar_2007 yang tadinya teks/string menjadi bilangan bulat dengan fungsi int(), kemudian mengalikan keduanya untuk menghitung luas dan hasilnya disimpan ke variabel luas_2007.
5. Baris 10 Mencetak teks "luas =" beserta nilai hasil perhitungan yang tersimpan pada variabel luas_2007 ke layar.

### 2.2.5 Konstanta_2611532007
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Buat file dengan nama Konstanta_2611532007.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variable ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_2007 = float(input('Masukkan nilai jari-jari: '))
luas_2007 = PI * jari_2007 * jari_2007
print("Luas lingkaran dengan jari jari %.2f adalah %.2f" % (jari_2007, luas_2007))
```

**Penjelasan :**
1. Baris 1-3 merupakan baris komentar yang membarikan intruksi tentang program .
2. Baris 4 Mengimpor penanda tipe (*type hint*) Final dari modul bawaan typing untuk menegaskan bahwa suatu variabel ditujukan sebagai konstanta (nilainya tidak boleh diubah).
3. Baris 5 Mendeklarasikan konstanta bernama PI bernilai 3.14 dan menandainya menggunakan : Final agar diperlakukan sebagai nilai tetap.
4. Baris 6 Mencetak nilai konstanta PI menggunakan format penulisan string legacy (%f), yang secara default akan menampilkan 6 angka di belakang koma (3.140000).
Meminta masukan (*input*) dari pengguna untuk nilai jari-jari, lalu langsung mengonversi teks masukan tersebut menjadi tipe bilangan desimal menggunakan fungsi float(), dan menyimpannya ke variabel jari_2007.
5. Baris 7 Menghitung luas lingkaran dengan rumus $\text{PI} \times r^2$ (PI * jari_2007 * jari_2007) dan menyimpannya ke dalam variabel luas_2007.
6. Baris 8 Mencetak hasil ke layar menggunakan format penulisan string (%.2f), di mana %.2f membatasi tampilan angka desimal variabel jari_2007 dan luas_2007 agar hanya memunculkan 2 angka di belakang koma.
